# Phase 4: Potato Leaf Disease Detection — Model Training & Evaluation

In this notebook, we build and train a custom Convolutional Neural Network (CNN) from scratch using TensorFlow and Keras.

### Pipeline Overview:
1. **Input Pipeline:** Load pre-split, augmented `tf.data` datasets
2. **Baseline CNN:** 4 Convolutional blocks + MaxPooling + GAP + Dense + Softmax
3. **Optimization:** Adam optimizer with Sparse Categorical Cross-Entropy
4. **Callbacks:** EarlyStopping, ModelCheckpoint, ReduceLROnPlateau
5. **Visualization:** Plot training vs. validation loss & accuracy curves

```python
import sys
from pathlib import Path

# Set project root
PROJECT_ROOT = Path("..").resolve()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

import tensorflow as tf
import matplotlib.pyplot as plt
import json

from src.config import CLASS_NAMES, NUM_CLASSES, IMAGE_SHAPE
from src.dataset import get_train_val_test_datasets
from src.model import build_baseline_cnn

print(f"TensorFlow Version: {tf.__version__}")
print(f"Target Classes ({NUM_CLASSES}): {CLASS_NAMES}")
```

```python
# Load Train, Validation, and strictly held-out Test datasets
train_ds, val_ds, test_ds = get_train_val_test_datasets(batch_size=32)

# Inspect Baseline CNN Architecture
model = build_baseline_cnn(input_shape=IMAGE_SHAPE, num_classes=NUM_CLASSES)
model.summary()
```

```python
# Load and Visualize Training History
history_path = PROJECT_ROOT / "data" / "processed" / "training_history_baseline.json"
with open(history_path, "r", encoding="utf-8") as f:
    history = json.load(f)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
epochs = history["epoch"]

ax1.plot(epochs, history["accuracy"], "o-", label="Train Accuracy", color="#2ecc71")
ax1.plot(epochs, history["val_accuracy"], "s--", label="Val Accuracy", color="#3498db")
ax1.set_title("Baseline CNN Accuracy")
ax1.set_xlabel("Epoch")
ax1.set_ylabel("Accuracy")
ax1.legend()
ax1.grid(True, linestyle="--", alpha=0.5)

ax2.plot(epochs, history["loss"], "o-", label="Train Loss", color="#e74c3c")
ax2.plot(epochs, history["val_loss"], "s--", label="Val Loss", color="#e67e22")
ax2.set_title("Baseline CNN Cross-Entropy Loss")
ax2.set_xlabel("Epoch")
ax2.set_ylabel("Loss")
ax2.legend()
ax2.grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```
