# Phase 2: PlantVillage Potato Leaf Dataset - Exploratory Data Analysis (EDA)

**Objective:** Verify raw image files, inspect class distributions, identify class imbalances, and confirm zero data leakage across stratified splits.

```python
import json
from pathlib import Path
from PIL import Image
import matplotlib.pyplot as plt

PROCESSED_DIR = Path("../data/processed")
with open(PROCESSED_DIR / "dataset_summary.json", "r") as f:
    summary = json.load(f)

print("Dataset Summary:")
print(json.dumps(summary, indent=2))
```

```python
# Display class distribution image generated during EDA
img = Image.open(PROCESSED_DIR / "eda_class_distribution.png")
plt.figure(figsize=(12, 6))
plt.imshow(img)
plt.axis("off")
plt.show()
```
